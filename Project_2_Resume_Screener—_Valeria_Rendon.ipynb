```python
!pip install -q -U google-genai

from google import genai
from google.colab import userdata

api_key = userdata.get("Gemini_API_KEY")
client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python
# Baseline Smart Resume Screener

def screen_resume(resume, job_description):
    prompt = f"""
You are a resume screening assistant. Compare the provided resume against the provided job description.
Rate the candidate's fit from 1-10 based only on job-related qualifications.
Ignore the candidate's name and gender when evaluating fit.

RESUME:
{resume}

JOB DESCRIPTION:
{job_description}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
resume = """
Technology Analyst | AI-Enabled Operations & Process Improvement

B.S. in Information Technology, expected Spring 2027.

5+ years of cross-industry experience supporting stakeholders,
coordinating time-sensitive information, maintaining accurate documentation,
and resolving operational issues.

AI & Analysis:
OpenAI/ChatGPT, Anthropic Claude, Google Gemini, generative AI,
prompt engineering fundamentals, LLM output validation,
document summarization, information extraction, workflow analysis.

Technology:
Microsoft 365, Google Workspace, ServiceNow, CCURE,
TCP/IP, DNS, DHCP, Java, C++.

Business & Operations:
Process improvement, stakeholder support, documentation, data accuracy,
cross-functional coordination, confidential records, client service,
and financial-services operations.

Experience includes:
- Accenture subcontractor supporting ServiceNow, workplace technology,
  secure access, and cross-functional operations.
- Administrative and healthcare operations experience.
- Wells Fargo banking experience involving financial services,
  security, compliance, auditing, transactions, and client support.
- Generative AI training and hands-on projects involving document
  summarization, information extraction, prompt design,
  process improvement, and output validation.
"""

job_description = """
MetLife - MII AI & Sustainability Analyst

The role supports AI and sustainability initiatives within MetLife
Investment Management.

Key responsibilities and qualifications include:
- Support the MII AI Lead and AI-related initiatives.
- Support reporting related to approved AI use and corporate AI goals.
- Work cross-functionally with technology teams and other departments.
- Explore AI and automation opportunities to improve workflows.
- Support sustainability reporting and improve information accessibility.
- Demonstrate comfort using AI tools.
- Apply analytical thinking, judgment, and attention to detail.
- Communicate and collaborate effectively across teams.
- Financial-services or investment-management experience is preferred.
- Business Administration, Accounting, or Finance education is preferred.
"""

print(screen_resume(resume, job_description))
```

Baseline Test 1 – MetLife MII AI & Sustainability Analyst

Result: 8/10

Observed Failure:
The baseline prompt produced a reasonable result, but it did not evaluate all major job requirements consistently. It emphasized the candidate's AI and financial-services background while overlooking the lack of direct sustainability reporting experience. It also treated familiarity with AI tools as stronger evidence of professional AI experience than the resume clearly demonstrates. The prompt needs clearer evaluation criteria and instructions to distinguish demonstrated experience from developing skills.

```python
# Improved Smart Resume Screener

def screen_resume_v2(resume, job_description):
    prompt = f"""
You are a resume screening assistant. Compare the resume against the job description.

Evaluate only job-related qualifications. Ignore the candidate's name and gender.

RULES:
- Use only evidence explicitly stated in the resume.
- Do not assume experience that is not documented.
- Distinguish professional experience from coursework, training, or developing skills.
- Consider both matching qualifications and important missing qualifications.
- Preferred qualifications should matter less than required qualifications.
- If the information is too incomplete or ambiguous to score reliably, state "Manual Review Required."

SCORING GUIDE:
9-10 = Excellent match; nearly all major qualifications demonstrated.
7-8 = Strong match; several major qualifications demonstrated with some gaps.
5-6 = Moderate match; relevant transferable qualifications but important gaps.
3-4 = Weak match; limited relevant qualifications.
1-2 = Very little demonstrated alignment.

EXAMPLES:
Example 1:
Candidate has professional AI experience, financial-services experience,
and direct sustainability reporting experience.
Result: 9/10

Example 2:
Candidate has AI training and financial-services experience but no
sustainability experience.
Result: 6/10

Example 3:
Candidate has general administrative experience but no demonstrated
AI, financial-services, or sustainability experience.
Result: 3/10

Return the result using this format:

FIT SCORE: X/10
MATCHING QUALIFICATIONS:
- ...
MISSING OR WEAK QUALIFICATIONS:
- ...
REASONING:
...
MANUAL REVIEW REQUIRED: Yes/No

RESUME:
{resume}

JOB DESCRIPTION:
{job_description}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
print(screen_resume_v2(resume, job_description))
```

### Prompt Iteration – Before vs. After

**Before:** The baseline prompt gave the candidate an 8/10. It recognized the candidate's AI and financial-services background, but overlooked the lack of sustainability experience and did not clearly distinguish AI training from professional AI experience.

**Change Made:** I added explicit scoring criteria, instructed the model to use only evidence stated in the resume, distinguish professional experience from training, identify missing qualifications, and use a Manual Review option for ambiguous cases. I also added few-shot examples to demonstrate different levels of candidate fit.

**After:** The improved prompt gave the candidate a 7/10. It still recognized the candidate's relevant AI, financial-services, analytical, and cross-functional skills, but now correctly identified the lack of sustainability experience and the difference between the candidate's IT education and the preferred business-related education.

**Conclusion:** The refined prompt produced a more structured and evidence-based evaluation.

### Technical Diligence

**Cost**

I used Gemini 3.5 Flash-Lite for my project because it is a lower-cost model that can handle this type of task.

For my estimate, I assumed that one resume screening would use around 2,000 input tokens and 300 output tokens.

For 1,000 resume screenings, the estimated cost would be about $1.35. The actual cost could be different depending on how long the resumes, job descriptions, and AI responses are.

**Responsible AI**

One concern with an AI resume screener is bias. I do not want the AI to rate someone differently because of their name or gender. To help prevent this, I instructed the model to ignore names and gender and focus only on qualifications related to the job.

Privacy is another concern because resumes can contain personal information such as phone numbers, email addresses, and home addresses. If this were a real system, unnecessary personal information should be removed or protected before sending the resume to the AI.

Finally, I would not use the AI score as the final hiring decision. The tool should help review a resume, but a person should still review the candidate, especially when the resume does not provide enough information.
